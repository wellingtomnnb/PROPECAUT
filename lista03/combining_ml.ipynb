```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import VotingClassifier, BaggingClassifier, RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Carregar a base de dados Wine
data = load_wine()
X, y = data.data, data.target

# Dividir os dados em treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Escalar os dados
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 1. Classificadores Individuais
# Logistic Regression
clf_lr = LogisticRegression(random_state=42, max_iter=1000)
clf_lr.fit(X_train, y_train)
y_pred_lr = clf_lr.predict(X_test)
accuracy_lr = accuracy_score(y_test, y_pred_lr)

# Support Vector Machine
clf_svc = SVC(probability=True, random_state=42)
clf_svc.fit(X_train, y_train)
y_pred_svc = clf_svc.predict(X_test)
accuracy_svc = accuracy_score(y_test, y_pred_svc)

# Decision Tree
clf_dt = DecisionTreeClassifier(random_state=42)
clf_dt.fit(X_train, y_train)
y_pred_dt = clf_dt.predict(X_test)
accuracy_dt = accuracy_score(y_test, y_pred_dt)

# 2. Classificadores Heterogêneos: Estratégia de Votação
voting_clf = VotingClassifier(
    estimators=[('lr', clf_lr), ('svc', clf_svc), ('dt', clf_dt)],
    voting='soft'  # Estratégia de votação com probabilidade
)

voting_clf.fit(X_train, y_train)
y_pred_voting = voting_clf.predict(X_test)
accuracy_voting = accuracy_score(y_test, y_pred_voting)

# 3. Classificadores Homogêneos: Bagging com Árvores de Decisão
bagging_clf = BaggingClassifier(
    estimator=DecisionTreeClassifier(random_state=42),
    n_estimators=10,
    random_state=42
)

bagging_clf.fit(X_train, y_train)
y_pred_bagging = bagging_clf.predict(X_test)
accuracy_bagging = accuracy_score(y_test, y_pred_bagging)

# 4. Classificadores Homogêneos: Random Forest
random_forest_clf = RandomForestClassifier(n_estimators=10, random_state=42)
random_forest_clf.fit(X_train, y_train)
y_pred_rf = random_forest_clf.predict(X_test)
accuracy_rf = accuracy_score(y_test, y_pred_rf)

# Exibir os resultados
print("Acurácia - Logistic Regression (Individual):", accuracy_lr)
print("Acurácia - SVM (Individual):", accuracy_svc)
print("Acurácia - Decision Tree (Individual):", accuracy_dt)
print("Acurácia - Voting (Heterogêneos):", accuracy_voting)
print("Acurácia - Bagging (Homogêneos):", accuracy_bagging)
print("Acurácia - Random Forest (Homogêneos):", accuracy_rf)
```

### COMBINAÇÃO DE CLASSIFICADORES
#### Definição e Teoria
1. Explique a diferença entre classificadores homogêneos e heterogêneos.

    Classificadores homogêneos são compostos por múltiplos modelos do mesmo tipo, enquanto classificadores heterogêneos utilizam modelos de tipos diferentes.


2. Qual é o principal benefício de usar classificadores combinados em comparação com classificadores individuais?
    
    O principal benefício de usar classificadores combinados é a maior precisão e robustez, pois eles podem compensar as fraquezas de modelos individuais, melhorando a generalização.

#### Prática e Implementação
1. Qual o objetivo de escalonar os dados antes de treinar os modelos? Cite exemplos de classificadores que se beneficiam dessa prática.

    O escalonamento visa ajustar a escala das variáveis, evitando que variáveis com magnitudes maiores dominem o modelo. Isso melhora a convergência e desempenho de algoritmos que dependem de distâncias ou gradientes.

2. No exemplo acima, por que foi usada a estratégia de votação com probabilidades ("soft voting") ao invés da estratégia de votação majoritária ("hard voting")?

    A estratégia de "soft voting" usa as probabilidades de cada classe para fazer a previsão final, levando em conta a confiança de cada classificador. Isso pode melhorar a precisão, especialmente quando os classificadores têm diferentes níveis de certeza. Já o "hard voting" apenas conta as classes preditas, ignorando a confiança de cada modelo.

#### Interpretação dos Resultados
1. Qual foi a estratégia que apresentou melhor desempenho no exemplo prático?
Justifique por que isso pode ter ocorrido.

    A estratégia de Voting. Isso provavelmente ocorreu porque a combinação de classificadores heterogêneos é mais robusta, compensando as fraquezas de cada classificador individual.

2. Como o número de estimadores influencia o desempenho do Bagging e da RandomForest? 
Cite possíveis limitações de aumentar indiscriminadamente esse número.

    Bagging e Random Forest melhoram o desempenho com mais estimadores, pois a agregação de múltiplos modelos reduz a variância e melhora a generalização.
    
    Limitações de aumentar o número de estimadores:

    Custo computacional: Mais estimadores exigem mais tempo de treinamento e maior consumo de memória.
    
    Diminuição de retornos: Após certo ponto, adicionar mais estimadores não melhora significativamente o desempenho e pode até aumentar o risco de overfitting, especialmente com conjuntos de dados pequenos.

#### Aplicações e Extensões
1. Escolha um problema do meu campo de pesquisa e proponha como diferentes estratégias de combinação de classificadores poderiam ser aplicadas.

    * Votação Heterogênea (Soft Voting): Para prever a demanda futura, você pode usar diferentes tipos de classificadores, como Regressão Logística, Árvore de Decisão e SVM, combinando as previsões de probabilidades de cada modelo. Isso pode melhorar a precisão ao equilibrar as vantagens de cada modelo para capturar diferentes aspectos dos dados de entrada.

    * Bagging (Homogêneo): Aplicando Bagging com Árvores de Decisão (ou outros classificadores de base), é possível reduzir a variância do modelo, especialmente quando há muita variabilidade nos dados históricos de demanda. A combinação de múltiplas árvores pode resultar em previsões mais estáveis e robustas, particularmente em cenários com muitos outliers ou dados ruidosos.

    * Random Forest (Homogêneo): Utilizando várias Árvores de Decisão com Random Forest pode ser útil para identificar padrões complexos na previsão de demanda, além de ajudar na redução de overfitting, mantendo o modelo generalizável e robusto ao realizar ajustes dinâmicos na cadeia de suprimentos.

    * Stacking: Utilizando diferentes modelos, como Redes Neurais e SVM, em um modelo de Stacking pode permitir que o modelo final aprenda a combinar as previsões de forma inteligente, aproveitando as diferentes forças de cada classificador. O meta-modelo (por exemplo, uma regressão logística) aprenderia a fazer a fusão das previsões de forma otimizada.

2. Descreva como adicionar mais diversidade entre classificadores poderia melhorar a robustez de um modelo combinado.
    
    Usar diferentes tipos de modelos permite que cada classificador enfoque diferentes aspectos dos dados e minimize a chance de erros em conjunto. Além de treinar os modelos em subconjuntos diferentes ou com diferentes transformações dos dados pode ajudar a capturar uma gama maior de padrões. Classificadores diversos também podem ser mais eficazes em compensar erros individuais, aumentando a precisão do modelo combinado. Por exemplo, um modelo de árvore de decisão pode ser bom em lidar com dados categóricos, enquanto um modelo SVM pode capturar relações complexas em dados numéricos. A combinação dessas abordagens reduz o risco de falhas, especialmente em cenários de previsão dinâmica e ajustes na cadeia de suprimentos. Isso resulta em um modelo mais generalizável, menos suscetível a overfitting e capaz de lidar com a complexidade e dinâmica de problemas reais na gestão de cadeias de suprimentos.
