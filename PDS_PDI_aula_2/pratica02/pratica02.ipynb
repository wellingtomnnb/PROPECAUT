### TAREFA 1 – LEITURA E ANÁLISE DAS ESTRUTURAS MATRICIAIS


```python
#importing libraries
import cv2 as cv
from matplotlib import pyplot as plt
```

```python
# a)  utilize   o   parâmetro cv2.IMREAD_UNCHANGED, de forma a carregar a imagem tal como ela 
# se encontra armazenada na memória

img_1 = cv.imread("./cameraman.tif", cv.IMREAD_UNCHANGED)
img_2 = cv.imread("./morangos.tif", cv.IMREAD_UNCHANGED)
```

```python
# b) exiba as imagens
# Exibir dimensões
print("Dimensão img_1:", img_1.shape)
print("Dimensão img_2:", img_2.shape)
```

```python
# OpenCV
cv.namedWindow("Imagem 1 - OpenCV", cv.WINDOW_AUTOSIZE)
cv.imshow("Imagem 1 - OpenCV", img_1)
cv.namedWindow("Imagem 2 - OpenCV", cv.WINDOW_AUTOSIZE)
cv.imshow("Imagem 2 - OpenCV", img_2)
cv.waitKey(10)
cv.destroyAllWindows()
```

```python
# Matplotlib
plt.figure("Imagem 1 - Matplotlib")
plt.imshow(img_1, cmap='gray')
plt.title("Imagem 1")
plt.axis('off')

plt.figure("Imagem 2 - Matplotlib")
plt.imshow(img_2)
plt.title("Imagem 2")
plt.axis('off')

plt.show()
```

### TAREFA 2 - SEPARANDO AS COMPONENTES (MATRIZES) DE CORES RGB

```python
# a) Canal vermelho (OpenCV carrega em BGR, então o canal vermelho é o índice 2)
verm_2 = img_2[:, :, 2]
```

```python
# b) Exibir apenas o canal vermelho
plt.figure("Morangos – Vermelho")
plt.imshow(verm_2, cmap='gray')
plt.title("Morangos – Vermelho")
plt.axis('off')
plt.show()
```

```python
# c) Separar canais verde e azul
verde_2 = img_2[:, :, 1]
azul_2 = img_2[:, :, 0]
```

```python
import numpy as np

# d) Reunir os canais na ordem RGB usando dstack
img_rgb = np.dstack((verm_2, verde_2, azul_2))
```

```python
# e) Exibir imagem RGB
plt.figure("Morangos – RGB")
plt.imshow(img_rgb)
plt.title("Morangos – RGB")
plt.axis('off')
plt.show()
```

### TAREFA 3 - CONVERTENDO UMA IMAGEM COLORIDA PARA:

```python
# a) Converter img_2 para tons de cinza
img_cinza = cv.cvtColor(img_2, cv.COLOR_BGR2GRAY)
```

```python
# b) Exibir imagem em tons de cinza
plt.figure("Morangos – Cinza")
plt.imshow(img_cinza, cmap='gray')
plt.title("Morangos – Cinza")
plt.axis('off')
plt.show()
```

```python
# c) Média dos canais manualmente
verm_2 = img_2[:, :, 2]
verde_2 = img_2[:, :, 1]
azul_2 = img_2[:, :, 0]
img_cinza_2 = ((verm_2.astype(np.float32) + verde_2.astype(np.float32) + azul_2.astype(np.float32)) / 3).astype(np.uint8)

plt.figure("Morangos – Cinza 2")
plt.imshow(img_cinza_2, cmap='gray')
plt.title("Morangos – Cinza 2")
plt.axis('off')
plt.show()
```

```python
# === HSV ===
# a) Converter img_2 para HSV
img_hsv = cv.cvtColor(img_2, cv.COLOR_BGR2HSV)

# b) Exibir imagem HSV
plt.figure("Morangos – HSV")
plt.imshow(cv.cvtColor(img_hsv, cv.COLOR_HSV2RGB))  # Para visualização correta
plt.title("Morangos – HSV")
plt.axis('off')
plt.show()
```

```python
# === BINÁRIA ===

# d) Converter img_1 para binária com limiar 128
_, img_bin = cv.threshold(img_2, 128, 255, cv.THRESH_BINARY)
```

```python
# e) Exibir imagem binária
plt.figure("Morangos – Binário")
plt.imshow(img_bin, cmap='gray')
plt.title("Morangos – Binário")
plt.axis('off')
plt.show()
```
